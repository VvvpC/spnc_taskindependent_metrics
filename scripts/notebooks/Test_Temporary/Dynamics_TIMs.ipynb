#### Grid search数据

```python
# 提取数据
with open('../Res_Tasks_Results/Reservoir_Res_beta20to50_theta001to06/results.pkl', 'rb') as f:
    grid_search = pickle.load(f)

# 定义名称
grid_MC = grid_search['runs'][0]['results']['MC']
grid_CQ = grid_search['runs'][1]['results']['CQ']
grid_nrmse = grid_search['runs'][2]['results']['NRMSE']
grid_acc = grid_search['runs'][3]['results']['acc']

grid_beta = grid_search['runs'][0]['results']['beta_prime']
grid_theta = grid_search['runs'][0]['results']['theta']

```

```python
# 插值法和heatmap函数
# 使用插值法平滑数据
from scipy.interpolate import griddata
import numpy as np
import pandas as pd

def interpolation(x, y, z, nx=200, ny=200):
    x = np.asarray(x).ravel()   
    y = np.asarray(y).ravel()
    z = np.asarray(z).ravel()

    # confirm the length of x and y are the same
    if x.shape[0] != y.shape[0] or x.shape[0] != z.shape[0]:
        raise ValueError("x and y must have the same length")
    
    # generate the grid
    xi = np.linspace(x.min(), x.max(), nx)
    yi = np.linspace(y.min(), y.max(), ny)
    Xi, Yi = np.meshgrid(xi, yi)

    # interpolate the data
    zi = griddata((x, y), z, (Xi, Yi), method='cubic')

    return Xi, Yi, zi

# 确保全局样式已应用
plot_style_config.set_pub_style()

def plot_heatmap(x, y, zi, 
                 xlabel=r'$\beta^{\prime}$', 
                 ylabel=r'Input rate ($\theta$/$T_0$)', 
                 value_label='MC', 
                 save_path=None):

    figsize = plot_style_config.get_figsize(width_mode='single', aspect_ratio=1.2)
    
    fig, ax = plt.subplots(figsize=figsize)
    
    pcm = ax.pcolormesh(x, y, zi, cmap='viridis', shading='auto', rasterized=True)
    

    cbar = plt.colorbar(pcm, ax=ax)
    
    cbar.set_label(value_label)
    
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    
    plt.tight_layout()
    
    # 6. 保存逻辑
    # 6. 保存逻辑：确保保存为SVG
    if save_path:
        # 检查 save_path 是否以 .svg 结尾，如果不是，则强制添加，确保保存为SVG格式
        if not save_path.lower().endswith('.svg'):
            # 如果路径存在但没有扩展名，我们添加 .svg
            # 如果路径已经有其他扩展名（如 .pdf），这会覆盖，但更安全的方式是在调用时传入正确的路径。
            # 为了简单，我们只在保存时指定格式为 'svg'
            pass 
            
        # 重点：plt.savefig 会根据文件扩展名自动确定格式，或使用 format='svg' 明确指定
        # 我们使用 format='svg' 来保证输出格式，并且设置 dpi=300 来保持非矢量元素的清晰度
        plt.savefig(save_path, format='svg', dpi=300) 
        print(f"Heatmap saved to {save_path} as SVG.")
        
    return fig
```

```python
# MC与dynamcis的关系
x = np.array(grid_beta)
y = np.array(grid_theta)
z = np.array(grid_MC)

Xi, Yi, zi = interpolation(x, y, z, nx=150, ny=150)

plot_heatmap(Xi, Yi, zi, value_label='Memory Capacity (MC)', save_path='MC_dynamics.svg')
```

```python
# CQ与dynamcis的关系
x = np.array(grid_beta)
y = np.array(grid_theta)
z = np.array(grid_CQ)

Xi, Yi, zi = interpolation(x, y, z, nx=150, ny=150)

plot_heatmap(Xi, Yi, zi, value_label='Computational Quality (CQ)', save_path='CQ_dynamics.svg')
```

```python
# nrmse与dynamcis的关系
x = np.array(grid_beta)
y = np.array(grid_theta)
z = np.array(grid_nrmse)

Xi, Yi, zi = interpolation(x, y, z, nx=100, ny=100)

plot_heatmap(Xi, Yi, zi, value_label='NRMSE', save_path='nrmse_dynamics.svg')
```

```python
# acc与dynamcis的关系
x = np.array(grid_beta)
y = np.array(grid_theta)
z = np.array(grid_acc)
z = (1-z)*100


Xi, Yi, zi = interpolation(x, y, z, nx=100, ny=100)

plot_heatmap(Xi, Yi, zi, value_label='Error rate(%)', save_path='acc_dynamics.svg')
```
